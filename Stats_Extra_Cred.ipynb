```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#making the initial random, uniform distribution
sample_num = 1000
distrib= np.random.uniform(1, 10, sample_num)
```

```python
#plot the distribution
plt.figure(figsize=(6,6))
plt.hist(distrib, bins=30, color='red', edgecolor='black', density=True)
plt.title('Initial Uniform Distribution')
plt.xlabel('1-10 Value')
plt.ylabel('Density of 1-10 Value')
```

```python
#pick two values from the random distribution and average the values together
#repeat this process 200 times

#store averages in empty array
averages = []

#loop to choose the values randomly 200 times
for _ in range(200):
    indices = np.random.choice(sample_num, 2, replace=False)
    selected_values = distrib[indices]
    avg = np.mean(selected_values)
    averages.append(avg)
```

```python
#plot the new values
plt.figure(figsize=(6,6))
plt.hist(averages, bins=30, color='red', edgecolor='black', density=True)
plt.title('New Averaged Values Distribution')
plt.xlabel('1-10 Value')
plt.ylabel('Density of 1-10 Value')
```

```python
#do this again
#lets pick and average 8 values
#store averages in empty array
averages_8 = []

#loop to choose the values randomly 200 times
for _ in range(200):
    indices = np.random.choice(sample_num, 8, replace=False)
    selected_values = distrib[indices]
    
    avg = np.mean(selected_values)
    averages_8.append(avg)
```

```python
#now plot these values
plt.figure(figsize=(6,6))
plt.hist(averages_8, bins=30, color='red', edgecolor='black', density=True)
plt.title('New Averaged Values Distribution')
plt.xlabel('1-10 Value')
plt.ylabel('Density of 1-10 Value')
```
